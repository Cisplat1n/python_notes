A list data type is an ordered sequence of elements that is normally comprised of strings,numbers, or even other lists. They are mutable and use a zero-based indexing (first element in a list starts at position 0)

```python
#basic list

cities = ["New York", "Los Angeles", "Chicago", "Houston", "Phoenix"]

#in order to access a certain element in the list, we can use its index
print(cities[0]) #Output: New York
print(cities[2]) #Output: Chicago
```

We can also use -ve indexing to access elements starting from the end of the list instead of the beginning. 

```python
print(cities[-1]) #Output: Phoenix (negative indexing starts from the end of the list)
```

Another handy way to create a list can be with the list constructor ( list() )

```python
developer = 'Jeff'

list(developer) #Output: ['J', 'e', 'f', 'f'] (a string can be treated as a list of characters  )
```

```python
#length of a list with the len() function
numbers = [1, 2, 3, 4, 5]
len(numbers) # 5 
```

```python
#since lists are mutable we can insert, delete, or modify elements in a list
programming_languages = ["Python", "Java", "Rust", "C++"]
print(programming_languages)
programming_languages[1] = "JavaScript" #modifying an element in the list
print(programming_languages) #Output: ['Python', 'JavaScript', 'Rust', '
```

```python
#we can also use del to remove elements from a list
del programming_languages[2] #removing the element at index 2 (Rust)
print(programming_languages) #Output: ['Python', 'JavaScript', 'C++']
```

```python
#it is also possible to determine if an element is in a list using the in keyword and receive a boolean value as a result
'Rust' not in programming_languages #Output: True
'Rust' in programming_languages #Output: False
'Python' in programming_languages #Output: True
```

We can also have nested lists within a list.

```python
developer = ['Alice', 25, ['Python', 'Rust', 'C++']]
developer[2] #Output: ['Python', 'Rust', 'C++'] (a list can contain other lists as elements, creating a nested list)

#we can also access elements in a nested list using multiple indices
developer[2][0] #Output: 'Python' (accessing the first element of the nested list)
```

Another useful technique with lists is unpacking its values. We are effectively able to assgin elements in a list to external variables.

```python
developer = ['Alice', 25, ['Python', 'Rust', 'C++']]

name,age, languages = developer #unpacking the list into separate variables
print(name) #Output: Alice
print(age) #Output: 25
print(languages) #Output: ['Python', 'Rust', 'C++']
```

```python
# it is also possible to unpack only some elements of the list and ignore the rest using the * operator
name,*rest = developer
print(name) #Output: Alice
print(rest) #Output: [25, ['Python', 'Rust', 'C++']]
```

If the number of variables on the left side of the assignmnet do not match the total number of elements in the list then an error will be raised.

```python
developer = ['Alice', 25, ['Python', 'Rust', 'C++']]
name,age,languages,city = developer #this will raise a ValueError because there are only 3 elements in the list but we are trying to unpack it into 4 variables
```

The final concept when it comes to lists is the slicing operation. Similar to strings, we can access certain elements from a list.

```python
desserts = ['Cake', 'Cookies', 'Ice Cream', 'Pie', 'Brownies']
desserts[1:4] #Output: ['Cookies', 'Ice Cream', 'Pie'] (slicing the list to get a sublist from index 1 to 3)

#ofcourse we can also select a step in the slicing
desserts[1::2] #Output: ['Cookies', 'Pie'] (slicing the list to get every second element starting from index 1)
```
